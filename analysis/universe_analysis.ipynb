```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Load Data
df = pd.read_csv("../universe.csv")  # .. because it's one folder up
print(f"Loaded {len(df)} stocks.")

# 2. Sector Pie Chart
plt.figure(figsize=(8, 8))
df['Sector'].value_counts().plot.pie(autopct='%1.1f%%', cmap='Pastel1')
plt.title("Universe Sector Exposure")
plt.ylabel("")
plt.show()

# 3. Top Industries Bar Chart
plt.figure(figsize=(10, 6))
df['Industry'].value_counts().head(15).plot(kind='barh', color='teal')
plt.title("Top 15 Industries")
plt.gca().invert_yaxis()
plt.show()
```
